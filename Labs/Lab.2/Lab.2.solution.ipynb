*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
winner_is_2 = [[2, 2, 0],
	[2, 1, 0],
	[2, 1, 1]]

winner_is_1 = [[1, 2, 0],
	[2, 1, 0],
	[2, 1, 1]]

winner_is_also_1 = [[0, 1, 0],
	[2, 1, 0],
	[2, 1, 1]]

no_winner = [[1, 2, 0],
	[2, 1, 0],
	[2, 1, 2]]

also_no_winner = [[1, 2, 0],
	[2, 1, 0],
	[2, 1, 0]]
````

````python
def check_game_state(matrix):
    n = len(matrix)

    for row in matrix:
        if len(set(row)) == 1 and row[0] != 0:
            return row[0]
            
    for col in range(n):
        column = [matrix[row][col] for row in range(n)]
        if len(set(column)) == 1 and column[0] != 0:
            return column[0]

    diag1 = [matrix[i][i] for i in range(n)]
    if len(set(diag1)) == 1 and diag1[0] != 0:
        return diag1[0]
        
    diag2 = [matrix[i][n - 1 - i] for i in range(n)]
    if len(set(diag2)) == 1 and diag2[0] != 0:
        return diag2[0]

    for row in matrix:
        if 0 in row:
            return -1
            
    return 0
````

````python
winner_x = [
    [1, 2, 0],
    [2, 1, 0],
    [2, 0, 1]
]

winner_o = [
    [1, 2, 1],
    [1, 2, 0],
    [0, 2, 0]
]

incomplete = [
    [1, 0, 0],
    [0, 2, 0],
    [0, 0, 0]
]

draw = [
    [1, 2, 1],
    [1, 2, 2],
    [2, 1, 1]
]

print(f"Winner X expected 1: {check_game_state(winner_x)}")
print(f"Winner O expected 2: {check_game_state(winner_o)}")
print(f"Incomplete expected -1: {check_game_state(incomplete)}")
print(f"Draw expected 0: {check_game_state(draw)}")
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
def make_move(board, player, row, col):

    if 0 <= row < len(board) and 0 <= col < len(board[0]):
        if board[row][col] == 0:
            board[row][col] = player
            return True          
    return False
````

````python
test_board = [[0, 0, 0], [0, 1, 0], [0, 0, 0]]
print(f"Valid move: {make_move(test_board, 2, 0, 0)}")
print(f"Taken spot: {make_move(test_board, 2, 1, 1)}")  
print(f"Out of bounds: {make_move(test_board, 1, 5, 5)}") 
````

*Exercise 6:* Modify Exercise 3 to show **row and column labels** so that players can specify locations like `"A2"` or `"C1"`.

````python
def draw_labeled_board(matrix):
    symbols = {0: " ", 1: "X", 2: "O"}
    n = len(matrix)
    col_headers = "   " + "   ".join([str(i + 1) for i in range(n)])
    print(col_headers)
    horizontal_line = "  " + " ---" * n
    
    for i, row in enumerate(matrix):
        print(horizontal_line)
        row_label = chr(ord('A') + i)
        row_str = f"{row_label} |"
        
        for cell in row:
            row_str += f" {symbols[cell]} |"
        print(row_str)
        
    print(horizontal_line)
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
def make_string_move(board, player, location):

    if len(location) < 2:
        return False
        
    row_char = location[0].upper()
    col_str = location[1:]
    
    try:
        row = ord(row_char) - ord('A')
        col = int(col_str) - 1
        return make_move(board, player, row, col)
    except ValueError:
        return False
````

````python
test_board_2 = [[0, 0], [0, 0]]
print(f"Move A2: {make_string_move(test_board_2, 1, 'A2')}")
print(f"Move B1: {make_string_move(test_board_2, 2, 'b1')}")
print(f"Move C5: {make_string_move(test_board_2, 1, 'C5')}")
````

*Exercise 8:* Write a function that is called with a board and player number, takes input from the player using Python's `input()`, and modifies the board using your function from Exercise 7.

Keep asking for input until the player enters a valid location that results in a valid move.


````python
def get_player_input(board, player):
    symbols = {1: "X", 2: "O"}
    
    while True:
        location = input(f"Player {symbols[player]}, enter your move (e.g., A2): ")
        if make_string_move(board, player, location):
            break
        else:
            print("Invalid move. The space is taken, out of bounds, or formatted poorly.")
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
def play_tic_tac_toe(n=3):
    board = create_board(n)
    current_player = 1
    symbols = {1: "X", 2: "O"}
    
    print(f"Starting a {n}x{n} game of Tic-Tac-Toe!")
    draw_labeled_board(board)
    
    while True:
        get_player_input(board, current_player)
        draw_labeled_board(board)
        status = check_game_state(board)
        
        if status != -1:
            if status == 0:
                print("Game Over: It's a draw!")
            else:
                print(f"Game Over: Player {symbols[status]} wins!")
            break
        current_player = 2 if current_player == 1 else 1
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
import random
def get_computer_move_simple(board, player):
    n = len(board)

    if player == 1:
        opponent = 2
    else:
        opponent = 1

    for r in range(n):
        for c in range(n):
            if board[r][c] == 0:
                board[r][c] = player
                if check_game_state(board) == player:
                    board[r][c] = 0
                    return r, c
                board[r][c] = 0

    for r in range(n):
        for c in range(n):
            if board[r][c] == 0: 
                board[r][c] = opponent
                if check_game_state(board) == opponent:
                    board[r][c] = 0
                    return r, c
                board[r][c] = 0

    empty_spots = []
    for r in range(n):
        for c in range(n):
            if board[r][c] == 0:
                empty_spots.append((r, c))
                
    if len(empty_spots) > 0:
        random_index = random.randint(0, len(empty_spots) - 1)
        return empty_spots[random_index]
        
    return None
````

````python
print("Testing Simple AI (Player 2 / 'O')...")

test_board_11 = [
    [1, 1, 0],
    [0, 2, 0],
    [0, 0, 0]
]

draw_labeled_board(test_board_11)

move_11 = get_computer_move_simple(test_board_11, 2)
print("Simple AI chooses:", move_11)
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
def get_score(board, current_player, maximizing_player, depth, max_depth):
    status = check_game_state(board)

    if status == maximizing_player:
        return 10 - depth
    elif status != 0 and status != -1: 
        return -10 + depth
    elif status == 0 or depth == max_depth:
        return 0

    if current_player == 1:
        next_player = 2
    else:
        next_player = 1

    n = len(board)

    if current_player == maximizing_player:
        best_score = -9999
        for r in range(n):
            for c in range(n):
                if board[r][c] == 0:
                    board[r][c] = current_player
                    score = get_score(board, next_player, maximizing_player, depth + 1, max_depth)
                    board[r][c] = 0
                    if score > best_score:
                        best_score = score
        return best_score

    else:
        best_score = 9999
        for r in range(n):
            for c in range(n):
                if board[r][c] == 0:
                    board[r][c] = current_player
                    score = get_score(board, next_player, maximizing_player, depth + 1, max_depth)
                    board[r][c] = 0
                    if score < best_score:
                        best_score = score
        return best_score

def get_computer_move_smart(board, player, max_depth=3):
    n = len(board)
    best_score = -9999
    best_r = -1
    best_c = -1
    
    if player == 1:
        next_player = 2
    else:
        next_player = 1

    for r in range(n):
        for c in range(n):
            if board[r][c] == 0:
                board[r][c] = player
                score = get_score(board, next_player, player, 1, max_depth)
                board[r][c] = 0
                if score > best_score:
                    best_score = score
                    best_r = r
                    best_c = c

    if best_r != -1:
        return best_r, best_c

    return get_computer_move_simple(board, player)
````

````python
print("Testing Smart AI (Player 2 / 'O')...")

test_board_12 = [
    [2, 2, 0],
    [1, 1, 0],
    [0, 0, 0]
]

draw_labeled_board(test_board_12)

move_12 = get_computer_move_smart(test_board_12, 2, max_depth=3)
print("Smart AI chooses:", move_12)
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
def simulate_ai_games(n, games, smart_depth):
    smart_wins = 0
    draws = 0

    for game_number in range(games):
        board = create_board(n)
        current_player = 1

        while True:
            if current_player == 1:
                move = get_computer_move_simple(board, 1)
            else:
                move = get_computer_move_smart(board, 2, max_depth=smart_depth)
            if move != None:
                row = move[0]
                col = move[1]
                make_move(board, current_player, row, col)
            status = check_game_state(board)
            if status != -1:
                if status == 2:
                    smart_wins = smart_wins + 1
                elif status == 0:
                    draws = draws + 1
                break

            if current_player == 1:
                current_player = 2
            else:
                current_player = 1

    print("Board size:", n, "x", n)
    print("Smart AI Wins:", smart_wins)
    print("Draws:", draws)
    print("Win percentage:", (smart_wins / games) * 100, "%")
    print("--------------------")
````

````python
print("Starting AI vs AI simulations...\n")

simulate_ai_games(n=3, games=10, smart_depth=4)
simulate_ai_games(n=4, games=10, smart_depth=2)
simulate_ai_games(n=5, games=10, smart_depth=2)
````
