8. Add a function to the object classes, including the base, that tests if a given set of $x$ and $y$ coordinates are inside of the object. You'll have to think through how to determine if a set of coordinates are inside an object for each object type.

```python
import math

class Shape:
    def __init__(self, x, y):
        self.__x = x
        self.__y = y

    def get_area(self):
        raise NotImplementedError("This method must be overridden by a subclass.")

    def get_perimeter(self):
        raise NotImplementedError("This method must be overridden by a subclass.")

    def is_inside(self, test_x, test_y):
        raise NotImplementedError("This method must be overridden by a subclass.")

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y


class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width

    def is_inside(self, test_x, test_y):
        x_min = self.get_x()
        x_max = self.get_x() + self.__length
        y_min = self.get_y()
        y_max = self.get_y() + self.__width
        
        if test_x >= x_min and test_x <= x_max and test_y >= y_min and test_y <= y_max:
            return True
        else:
            return False


class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius

    def is_inside(self, test_x, test_y):
        center_x = self.get_x()
        center_y = self.get_y()
        
        distance_squared = (test_x - center_x) ** 2 + (test_y - center_y) ** 2
        radius_squared = self.__radius ** 2
        
        if distance_squared <= radius_squared:
            return True
        else:
            return False


class Triangle(Shape):
    def __init__(self, side_a, side_b, side_c, x, y):
        super().__init__(x, y)
        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c

    def is_inside(self, test_x, test_y):
        x1 = self.get_x()
        y1 = self.get_y()
        
        x2 = x1 + self.__side_a
        y2 = y1
        
        x3 = x1 + (self.__side_a**2 + self.__side_c**2 - self.__side_b**2) / (2 * self.__side_a)
        y3 = y1 + math.sqrt(self.__side_c**2 - (x3 - x1)**2)
        
        def calc_area(px1, py1, px2, py2, px3, py3):
            area = 0.5 * abs(px1 * (py2 - py3) + px2 * (py3 - py1) + px3 * (py1 - py2))
            return area
        
        total_area = calc_area(x1, y1, x2, y2, x3, y3)
        
        area1 = calc_area(test_x, test_y, x1, y1, x2, y2)
        area2 = calc_area(test_x, test_y, x2, y2, x3, y3)
        area3 = calc_area(test_x, test_y, x3, y3, x1, y1)
        
        sum_of_areas = area1 + area2 + area3
        
        tolerance = 0.00001
        difference = abs(total_area - sum_of_areas)
        
        if difference <= tolerance:
            return True
        else:
            return False


#Example
my_rect = Rectangle(length=10, width=5, x=0, y=0)
print(f"Is (5, 2) inside rectangle? {my_rect.is_inside(5, 2)}")
print(f"Is (15, 2) inside rectangle? {my_rect.is_inside(15, 2)}")

my_circle = Circle(radius=5, x=0, y=0)
print(f"\nIs (2, 2) inside circle? {my_circle.is_inside(2, 2)}")
print(f"Is (6, 6) inside circle? {my_circle.is_inside(6, 6)}")
```

9. Add a function in the base class of the object classes that returns true/false testing that the object overlaps with another object.

```python
import math

class Shape:
    def __init__(self, x, y):
        self.__x = x
        self.__y = y

    def get_area(self):
        raise NotImplementedError("This method must be overridden by a subclass.")

    def get_perimeter(self):
        raise NotImplementedError("This method must be overridden by a subclass.")

    def get_perimeter_points(self):
        raise NotImplementedError("This method must be overridden by a subclass.")

    def is_inside(self, test_x, test_y):
        raise NotImplementedError("This method must be overridden by a subclass.")

    def overlaps(self, other_shape):
        my_points = self.get_perimeter_points()

        for point in my_points:
            point_x = point[0]
            point_y = point[1]
            if other_shape.is_inside(point_x, point_y) == True:
                return True

        other_points = other_shape.get_perimeter_points()

        for point in other_points:
            point_x = point[0]
            point_y = point[1]
            if self.is_inside(point_x, point_y) == True:
                return True

        return False

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y


class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width

    def is_inside(self, test_x, test_y):
        x_min = self.get_x()
        x_max = self.get_x() + self.__length
        y_min = self.get_y()
        y_max = self.get_y() + self.__width
        
        if test_x >= x_min and test_x <= x_max and test_y >= y_min and test_y <= y_max:
            return True
        else:
            return False

    def get_perimeter_points(self):
        points = []
        x = self.get_x()
        y = self.get_y()
        
        points.append((x, y))
        points.append((x + self.__length, y))
        points.append((x + self.__length, y + self.__width))
        points.append((x, y + self.__width))
        
        return points

class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius

    def is_inside(self, test_x, test_y):
        center_x = self.get_x()
        center_y = self.get_y()
        
        distance_squared = (test_x - center_x) ** 2 + (test_y - center_y) ** 2
        radius_squared = self.__radius ** 2
        
        if distance_squared <= radius_squared:
            return True
        else:
            return False

    def get_perimeter_points(self):
        points = []
        for i in range(16):
            angle = i * (2 * math.pi / 16)
            point_x = self.get_x() + self.__radius * math.cos(angle)
            point_y = self.get_y() + self.__radius * math.sin(angle)
            points.append((point_x, point_y))
        return points


#Example
rect1 = Rectangle(length=10, width=5, x=0, y=0)

circle1 = Circle(radius=2, x=10, y=5)

circle2 = Circle(radius=2, x=20, y=20)

print(f"Does rect1 overlap with circle1? {rect1.overlaps(circle1)}") 
print(f"Does rect1 overlap with circle2? {rect1.overlaps(circle2)}")
```

```python
import math

class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.data = []
        for i in range(height):
            row = []
            for j in range(width):
                row.append(' ')
            self.data.append(row)

    def set_pixel(self, row, col, char='*'):
        if row >= 0:
            if row < self.height:
                if col >= 0:
                    if col < self.width:
                        self.data[row][col] = char

    def get_pixel(self, row, col):
        return self.data[row][col]
    
    def clear_canvas(self):
        for i in range(self.height):
            for j in range(self.width):
                self.data[i][j] = ' '
    
    def v_line(self, x, y, h, **kargs):
        for i in range(x, x + h):
            self.set_pixel(i, y, **kargs)

    def h_line(self, x, y, w, **kargs):
        for i in range(y, y + w):
            self.set_pixel(x, i, **kargs)
            
    def line(self, x1, y1, x2, y2, **kargs):
        if y2 - y1 != 0:
            slope = (x2 - x1) / (y2 - y1)
            for y in range(y1, y2):
                x = x1 + int(slope * (y - y1))
                self.set_pixel(x, y, **kargs)
            
    def display(self):
        for row in self.data:
            line_str = ""
            for char in row:
                line_str = line_str + char
            print(line_str)


class Shape:
    def __init__(self, x, y):
        self.__x = x
        self.__y = y

    def get_area(self):
        raise NotImplementedError()

    def get_perimeter(self):
        raise NotImplementedError()

    def get_perimeter_points(self):
        raise NotImplementedError()

    def is_inside(self, test_x, test_y):
        raise NotImplementedError()

    def paint(self, canvas, char="*"):
        for row in range(canvas.height):
            for col in range(canvas.width):
                if self.is_inside(col, row) == True:
                    canvas.set_pixel(row, col, char)

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y


class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width

    def is_inside(self, test_x, test_y):
        x_min = self.get_x()
        x_max = self.get_x() + self.__length
        y_min = self.get_y()
        y_max = self.get_y() + self.__width
        
        if test_x >= x_min and test_x <= x_max and test_y >= y_min and test_y <= y_max:
            return True
        else:
            return False

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width


class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius

    def is_inside(self, test_x, test_y):
        center_x = self.get_x()
        center_y = self.get_y()
        
        distance_squared = (test_x - center_x) ** 2 + (test_y - center_y) ** 2
        radius_squared = self.__radius ** 2
        
        if distance_squared <= radius_squared:
            return True
        else:
            return False


class Triangle(Shape):
    def __init__(self, side_a, side_b, side_c, x, y):
        super().__init__(x, y)
        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c

    def is_inside(self, test_x, test_y):
        x1 = self.get_x()
        y1 = self.get_y()
        x2 = x1 + self.__side_a
        y2 = y1
        x3 = x1 + (self.__side_a**2 + self.__side_c**2 - self.__side_b**2) / (2 * self.__side_a)
        y3 = y1 + math.sqrt(self.__side_c**2 - (x3 - x1)**2)
        
        def calc_area(px1, py1, px2, py2, px3, py3):
            return 0.5 * abs(px1 * (py2 - py3) + px2 * (py3 - py1) + px3 * (py1 - py2))
        
        total_area = calc_area(x1, y1, x2, y2, x3, y3)
        area1 = calc_area(test_x, test_y, x1, y1, x2, y2)
        area2 = calc_area(test_x, test_y, x2, y2, x3, y3)
        area3 = calc_area(test_x, test_y, x3, y3, x1, y1)
        
        sum_of_areas = area1 + area2 + area3
        tolerance = 0.00001
        
        if abs(total_area - sum_of_areas) <= tolerance:
            return True
        else:
            return False


class CompoundShape(Shape):
    def __init__(self):
        super().__init__(0, 0)
        self.shapes = []

    def add_shape(self, shape):
        self.shapes.append(shape)

    def is_inside(self, test_x, test_y):
        for shape in self.shapes:
            if shape.is_inside(test_x, test_y) == True:
                return True
        return False

    def paint(self, canvas, char="*"):
        for shape in self.shapes:
            shape.paint(canvas, char)


if __name__ == "__main__":
    my_canvas = Canvas(width=40, height=20)

    house = CompoundShape()
    house_base = Rectangle(length=14, width=6, x=5, y=10)
    house_roof = Triangle(side_a=16, side_b=10, side_c=10, x=4, y=4)
    house.add_shape(house_base)
    house.add_shape(house_roof)

    sun = Circle(radius=3, x=32, y=5)
    ground = Rectangle(length=40, width=3, x=0, y=17)

    ground.paint(my_canvas, char="=")
    house.paint(my_canvas, char="#")
    sun.paint(my_canvas, char="O")

    my_canvas.display()
```

11. Create a `RasterDrawing` class. Demonstrate that you can create a drawing made of several shapes, paint the drawing, modify the drawing, and paint it again. 

```python
class RasterDrawing:
    def __init__(self):
        self.shapes = dict()
        self.shape_names = list()
        
    def add_shape(self, shape):
        if hasattr(shape, 'name') == False:
            shape.name = ""
            
        if shape.name == "":
            shape.name = self.assign_name()
            
        self.shapes[shape.name] = shape
        self.shape_names.append(shape.name)

    def update(self, canvas):
        canvas.clear_canvas()
        self.paint(canvas)
        
    def paint(self, canvas):
        for shape_name in self.shape_names:
            self.shapes[shape_name].paint(canvas)
            
    def assign_name(self):
        name_base = "shape"
        name = name_base + "_0"
        
        i = 1
        while name in self.shapes:
            name = name_base + "_" + str(i)
            i = i + 1
            
        return name



if __name__ == "__main__":
    class MutableRectangle(Rectangle):
        def set_length(self, new_length):
            self._Rectangle__length = new_length

    my_canvas = Canvas(width=40, height=15)
    my_drawing = RasterDrawing()

    box = MutableRectangle(length=10, width=5, x=5, y=5)
    ground = MutableRectangle(length=40, width=2, x=0, y=12)

    my_drawing.add_shape(box)
    my_drawing.add_shape(ground)

    print("Initial Drawing")
    my_drawing.paint(my_canvas)
    my_canvas.display()

    print("\nModifying the Drawing (Lengthening the Box)")
    target_shape = my_drawing.shapes["shape_0"]
    target_shape.set_length(25)

    my_drawing.update(my_canvas)
    my_canvas.display()
```

```python
class foo:
    def __init__(self,a,b=None):
        self.a=a
        self.b=b
        
    def __repr__(self):
        return "foo("+repr(self.a)+","+repr(self.b)+")"
    
    def save(self,filename):
        f=open(filename,"w")
        f.write(self.__repr__())
        f.close()
        
   
def foo_loader(filename):
    f=open(filename,"r")
    tmp=eval(f.read())
    f.close()
    return tmp
```
